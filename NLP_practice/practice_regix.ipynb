```python
string= r"sara was able to find the the item she was looking for"
new_string = re.sub("sara" , "sarah", string)
print(new_string)
```

```python
customer_reviews = [
    "sam was a great help to me in the store",
    "the cashier was very rude to me, I think her name was eleanor",
    "amazing work from sadeen!",
    "sarah was able to help me find the items I needed quickly",
    "lucy is such a great addition to the team",
    "great service from sara she found me what i wanted"
]
```

```python
sarah_reviews = []
pattern_to_find = r"sarah?"
```

```python
for review in customer_reviews:
    if re.search(pattern_to_find, review):
        sarah_reviews.append(review)

print(sarah_reviews)
```

```python
a_review = []
pattern_to_find = r"^a"
for review in customer_reviews:
    if re.search(pattern_to_find, review):
        a_review.append(review)

print(a_review)
```

```python
needed_reviews = []
pattern_to_find = r"(need|want)ed"
for review in customer_reviews:
    if re.search(pattern_to_find, review):
        needed_reviews.append(review)
print(needed_reviews)
```

```python
no_puinctuation_reviews = []
pattern_to_find = r"[^\w\s]"

for review in customer_reviews:
    if not re.search(pattern_to_find, review):
        no_puinctuation_reviews.append(review)

print(no_puinctuation_reviews)
```
